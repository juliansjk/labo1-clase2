```python
import numpy as np
import matplotlib.pyplot as plt

datos = np.loadtxt("1periodo.txt", skiprows=1)

datos_cortados = datos[datos > 0.8]
# datos_cortados = datos_cortados[datos_cortados < 1.26]

plt.figure(figsize=(8,4),dpi=150) #va primero

plt.subplot(1,2,1)
plt.hist(datos_cortados, bins=11, histtype="step", density=True, color="b", label="200 mediciones")


# plt.rcParams["axes.spines.top"] = True
# plt.rcParams["axes.spines.right"] = True

plt.title("Medición de 1 periodo",fontdict={'fontname':'Arial', 'fontsize':12})
plt.axline( (1.17 , 0) , (1.17 , 10), c='b')
plt.axline( (.885 , 0) , (.885 , 10), c='r')
plt.axline( (1.373 , 0) , (1.373 , 10), c='r')
plt.ylabel("Cantidad de datos" ,fontdict={'fontname':'Arial'})
plt.xlabel("Tiempo [s]" ,fontdict={'fontname':'Arial'})

plt.subplot(1,2,2)
plt.hist(datos_cortados[0:50], bins=9, histtype="step", density=True, color="r")


plt.title("1 periodo, 50 samples",fontdict={'fontname':'Arial','fontsize':12})
# plt.ylabel("Cantidad de datos" ,fontdict={'fontname':'Arial'})
plt.xlabel("Tiempo [s]" ,fontdict={'fontname':'Arial'})
plt.axline( (1.13 , 0) , (1.13 , 10), c='r')
# plt.xlim([1.05,1.2])


# plt.xticks([1.1])     manualmente decidis donde se marca el axe
# plt.savefig('nombre.png',dpi=300)
```

```python
datos_2 = np.loadtxt("datos_p2.txt",delimiter=" ", skiprows=1)

new_datos_2 = []
for x in datos_2:
    if(x < 2.38): new_datos_2.append(x -1.1)

plt.figure(figsize=(8,4),dpi=150) #va primero
plt.hist(new_datos_2, bins=8, histtype="step", density=True, color="r", label='1 periodo')
plt.hist(datos_cortados, bins=11, histtype="step", density=True, color="b", label='2 periodos')
plt.title("1 y 2 periodos (offset -1.1s)")
plt.axline( (1.15 , 0) , (1.15 , 10), c='b')
plt.axline( (1.1 , 0) , (1.1 , 10), c='r')
plt.xlabel("Tiempo [s]")
plt.ylabel("Cantidad de datos")
plt.legend()
```

```python
data_9p = np.loadtxt("9periodos.txt",delimiter=" ", skiprows=1)

data_9p_cut = data_9p[data_9p > 10.10]
data_9p_cut = data_9p_cut[data_9p_cut < 10.4]

data_9p_cut_offset = []

for x in data_9p:
    data_9p_cut_offset.append(x-9.1)

plt.figure(figsize=(8,4),dpi=150) 
plt.hist(data_9p_cut_offset, bins=8, histtype="step", density=True, color='r', label="9 periodos")
plt.hist(datos_cortados, bins=9, histtype="step", density=True, color="b", label="1 periodo")

plt.title("9 periodos (offset 9.1s) vs 1 periodo",fontdict={'fontname':'Arial', 'fontsize':12})
plt.ylabel("Cantidad de datos" ,fontdict={'fontname':'Arial'})
plt.xlabel("Tiempo [s]" ,fontdict={'fontname':'Arial'})
plt.legend()
```

```python
rawData = np.loadtxt('tiempoRespuesta.txt', skiprows=1)
newData = []
newDataOffset = []

cutData = rawData[rawData > 0.195]
cutData = cutData[cutData < 0.4]



plt.figure(figsize=(14,4),dpi=150) 

plt.subplot(1,3,1)
plt.hist(cutData, bins=12, histtype="step")
plt.ylim([0,37])

plt.title('Tiempo de respuesta')
plt.ylabel("Cantidad de datos" ,fontdict={'fontname':'Arial'})
plt.xlabel("Tiempo [s]" ,fontdict={'fontname':'Arial'})


index = 0
for x in cutData:
    if index != 0:
        newData.append(x - cutData[index-1])
        newDataOffset.append(x - cutData[index-1] + .101)
    index += 1

plt.subplot(1,3,2)
# plt.hist(cutData, bins="auto", histtype="step")
plt.hist(newData, bins=10, histtype="step")
plt.ylim([0,37])

# plt.axline( (0 , 4) , (.25 , 4), c='r')
plt.title('Tiempo de respuesta de tipo [ t1-t2, t3-t4, ...]')
plt.xlabel("Tiempo [s]" ,fontdict={'fontname':'Arial'})

plt.subplot(1,3,3)
# plt.hist(cutData, bins="auto", histtype="step")
plt.hist(newDataOffset, bins=10, histtype="step")
plt.ylim([0,37])

# plt.axline( (0 , 4) , (.25 , 4), c='r')
plt.title('De tipo [ t1-t2, t3-t4, ...] con offset')
plt.xlabel("Tiempo [s]" ,fontdict={'fontname':'Arial'})

plt.legend()
```

```python
index = 0
for x in cutData:
    cutData[index] = x + 0.85

    index += 1

index2 = 0
for x in newData:
    newData[index2] = x + 1.12

    index2 += 1

plt.figure(figsize=(8,4),dpi=150) 
plt.subplot(1,2,1)
plt.hist(cutData, bins=8, histtype="step", density=True, color='r', label="Tiempo respuesta")
plt.hist(datos_cortados, bins=9, histtype="step", density=True, color="b", label="1 periodo")
plt.ylim([0,14])


plt.title("1 periodo y tiempo de respuesta",fontdict={'fontname':'Arial', 'fontsize':12})
plt.ylabel("Cantidad de datos" ,fontdict={'fontname':'Arial'})
plt.xlabel("Tiempo [s]" ,fontdict={'fontname':'Arial'})

plt.subplot(1,2,2)
plt.hist(newData, bins=8, histtype="step", density=True, color='r', label="Tiempo respuesta")
plt.hist(datos_cortados, bins=9, histtype="step", density=True, color="b", label="1 periodo")
plt.ylim([0,14])


plt.title("1 periodo y [ t1-t2, t3-t4, ...]",fontdict={'fontname':'Arial', 'fontsize':12})
plt.xlabel("Tiempo [s]" ,fontdict={'fontname':'Arial'})
```
