```python
import numpy as np
import matplotlib.pyplot as plt

datos = np.loadtxt("1periodo.txt", skiprows=1)

datos_cortados = datos[datos > 1.05]
datos_cortados = datos_cortados[datos_cortados < 1.26]

plt.figure(figsize=(8,4),dpi=150) #va primero

plt.subplot(1,2,1)
plt.hist(datos_cortados, bins=11, histtype="step", density=True, color="b", label="200 mediciones")

plt.title("1 periodo, 200 samples",fontdict={'fontname':'Arial', 'fontsize':12})
plt.axline( (1.15 , 0) , (1.15 , 10), c='b')
plt.ylabel("Cantidad de datos" ,fontdict={'fontname':'Arial'})
plt.xlabel("Tiempo [s]" ,fontdict={'fontname':'Arial'})

plt.subplot(1,2,2)
plt.hist(datos_cortados[0:50], bins=9, histtype="step", density=True, color="r")


plt.title("1 periodo, 50 samples",fontdict={'fontname':'Arial','fontsize':12})
# plt.ylabel("Cantidad de datos" ,fontdict={'fontname':'Arial'})
plt.xlabel("Tiempo [s]" ,fontdict={'fontname':'Arial'})
plt.axline( (1.15 , 0) , (1.15 , 10), c='r')
# plt.xlim([1.05,1.2])


# plt.xticks([1.1])     manualmente decidis donde se marca el axe
# plt.savefig('nombre.png',dpi=300)
```

```python
datos_2 = np.loadtxt("datos_p2.txt",delimiter=" ", skiprows=1)

new_datos_2 = []
for x in datos_2:
    if(x < 2.38): new_datos_2.append(x -1.1)

plt.hist(new_datos_2, bins=8, histtype="step", density=True, color="r", label='1 periodo')
plt.hist(datos_cortados, bins=11, histtype="step", density=True, color="b", label='2 periodos')
plt.title("1 y 2 periodos (offset -1.1s)")
plt.axline( (1.15 , 0) , (1.15 , 10), c='b')
plt.axline( (1.1 , 0) , (1.1 , 10), c='r')
plt.xlabel("Tiempo [s]")
plt.ylabel("Cantidad de datos")
plt.legend()
```

```python
data_9p = np.loadtxt("9periodos.txt",delimiter=" ", skiprows=1)

data_9p_cut = data_9p[data_9p > 10.10]
data_9p_cut = data_9p_cut[data_9p_cut < 10.4]

data_9p_cut_offset = []

for x in data_9p_cut:
    data_9p_cut_offset.append(x-9.1)

plt.figure(figsize=(8,4),dpi=150) 
plt.hist(data_9p_cut_offset, bins=8, histtype="step", density=True, color='r', label="9 periodos")
plt.hist(datos_cortados, bins=11, histtype="step", density=True, color="b", label="1 periodo")

plt.title("9 periodos (offset 9.1s) vs 1 periodo",fontdict={'fontname':'Arial', 'fontsize':12})
plt.ylabel("Cantidad de datos" ,fontdict={'fontname':'Arial'})
plt.xlabel("Tiempo [s]" ,fontdict={'fontname':'Arial'})
plt.legend()
```

```python
rawData = np.loadtxt('tiempoRespuesta.txt', skiprows=1)
newData = []

index = 0
for x in rawData:
    if index != 0:
        newData.append(x - rawData[index-1])
    index += 1

plt.hist(rawData, bins="auto", histtype="step")
plt.hist(newData, bins="auto", histtype="step")
plt.title('Tiempo de respuesta y tiempo respuesta restado')
```
