```python
import numpy as np
import matplotlib.pyplot as plt

datos = np.loadtxt("1periodo.txt", skiprows=1)


b = 20


datos_cortados = datos[datos > 1]
datos_cortados = datos_cortados[datos_cortados < 1.26]

plt.hist(datos_cortados, bins=b, histtype="step", density=True, color="b", label="200 mediciones")
plt.hist(datos_cortados[0:50], bins=b, histtype="step", density=True, color="r")
#plt.hist(new_datos_2, bins = b, histtype = "step", density = True, color = "g")
plt.title("Un periodo, 200 y 50 samples")
plt.xlabel("Tiempo [s]")
plt.ylabel("Cantidad de datos")
plt.legend()
plt.xlim([0.8,1.4])
```

```python
datos_2 = np.loadtxt("datos_p2.txt",delimiter=" ", skiprows=1)

new_datos_2 = []
for x in datos_2:
    new_datos_2.append(x- 1.1)

plt.hist(datos_2, bins=b, histtype="step", density=True, color="b")
plt.hist(datos_cortados, bins=b, histtype="step", density=True, color="b")
plt.hist(new_datos_2, bins = b, histtype = "step", density = True, color = "g")
plt.title("Comparación entre 1 periodo y 2 periodos menos 1.1s")
plt.xlabel("Tiempo [s]")
plt.ylabel("Cantidad de datos")
#plt.xlim([0.8,1.4])
```

```python
datos_T_2 = np.loadtxt("9periodos.txt",delimiter=" ", skiprows=1)

datos_T_2cortados = datos_T_2[datos_T_2 > 10.10]
datos_T_2cortados = datos_T_2cortados[datos_T_2cortados < 10.4]

plt.hist(datos_T_2cortados, bins="auto", histtype="step")
plt.title("Medicion de 9 periodos consecutivos")
plt.xlabel("Tiempo [s]")
plt.ylabel("Cantidad de datos")
```

```python
rawData = np.loadtxt('tiempoRespuesta.txt', skiprows=1)
newData = []

index = 0
for x in rawData:
    if index != 0:
        newData.append(x - rawData[index-1])
    index += 1

plt.hist(rawData, bins="auto", histtype="step")
plt.hist(newData, bins="auto", histtype="step")
plt.title('Tiempo de respuesta y tiempo respuesta restado')
```
